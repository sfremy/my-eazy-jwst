```python
from astropy.coordinates import SkyCoord
import astropy.units as u

def find_match_by_coor(ra_target, dec_target, catalog):
    """
    Finds the closest matching target in the msaexp catalog by RA and Dec,
    and returns its S3 bucket folder root and spec.fits filename.
    """
    # 1. Convert the entire catalog coordinates into SkyCoord structures
    catalog_coords = SkyCoord(ra=catalog['ra']*u.deg, dec=catalog['dec']*u.deg)
    target_coord = SkyCoord(ra=ra_target*u.deg, dec=dec_target*u.deg)
    
    # 2. Match coordinates (finds the absolute closest index)
    idx, d2d, _ = target_coord.match_to_catalog_sky(catalog_coords)
    
    # Extract the matching row
    match_row = catalog[idx]
    
    if d2d.arcsec[0] > 1.0:
        return None
        
        
    return match_row
```

#### Emission Line Fitting
- Linear continuum
- Two-component Gaussian for Balmer lines
- Single-component for [OIII] 4959/5007 doublet, OI 6300 & OI 8446
- NO [NII] doublet
- Balmer + OIII tied together
- OIII doublet flux ratio 1:2.98

Priors are in Hviding et al. (2025) except for broadline FWHM which is set to uniform distribution in range (FWHMnarrow + 100, 5000km/s)
OI prior set to [0, 2000km/s]

Also fit the non-LRD Hviding et al. emitters (37 RUBIES sources), NII doublet fixed @ 1:2.95

```python
import jax.numpy as jnp
import numpyro
import numpyro.distributions as dist

C_KMS = 2.99792458e5  # km/s


def gaussian_profile(lam, lam_0, flux, sigma_v):
    # Convert velocity dispersion (km/s) to wavelength width (Angstroms)
    sigma_lam = lam_0 * (sigma_v / C_KMS)
    norm = flux / (sigma_lam * jnp.sqrt(2.0 * jnp.pi))
    return norm * jnp.exp(-0.5 * ((lam - lam_0) / sigma_lam) ** 2)


def joint_emission_lines(lam_rest, flux, flux_err):
    # -------------------------------------------------------------
    # 1. Tied Kinematics (in km/s)
    # -------------------------------------------------------------
    # Narrow component velocity dispersion (NLR: shared by narrow Balmer & [OIII])
    sigma_narrow = numpyro.sample(
        "sigma_narrow",
        dist.Uniform(low=100.0, high=800.0),  # PRISM line-spread limited
    )

    # Broad component velocity dispersion (BLR: shared by broad Balmer)
    sigma_broad = numpyro.sample(
        "sigma_broad", dist.Uniform(low=1200.0, high=6000.0)
    )

    # Small systematic velocity offset/redshift residual
    v_off = numpyro.sample("v_off", dist.Normal(loc=0.0, scale=300.0))

    # Doppler shift the rest wavelengths by v_off
    lam_shift = lambda lam0: lam0 * (1.0 + v_off / C_KMS)

    # -------------------------------------------------------------
    # 2. Line Flux Amplitudes (strictly positive)
    # -------------------------------------------------------------
    # Balmer line fluxes
    flux_ha_broad = numpyro.sample(
        "flux_ha_broad", dist.HalfNormal(scale=10.0)
    )
    flux_ha_narrow = numpyro.sample(
        "flux_ha_narrow", dist.HalfNormal(scale=5.0)
    )
    flux_hb_broad = numpyro.sample(
        "flux_hb_broad", dist.HalfNormal(scale=5.0)
    )
    flux_hb_narrow = numpyro.sample(
        "flux_hb_narrow", dist.HalfNormal(scale=2.0)
    )

    # [OIII] doublet: sample the main 5007 line, fix 4959 to 1:2.98
    flux_oiii_5007 = numpyro.sample(
        "flux_oiii_5007", dist.HalfNormal(scale=5.0)
    )
    flux_oiii_4959 = flux_oiii_5007 / 2.98

    # [O I] 8446 (single Gaussian, usually tied to narrow or sampled independently)
    flux_oi_8446 = numpyro.sample("flux_oi_8446", dist.HalfNormal(scale=2.0))

    # -------------------------------------------------------------
    # 3. Sum All Components
    # -------------------------------------------------------------
    # Rest vacuum wavelengths
    l_ha = lam_shift(6564.61)
    l_hb = lam_shift(4862.68)
    l_oiii_4959 = lam_shift(4960.30)
    l_oiii_5007 = lam_shift(5008.24)
    l_oi_8446 = lam_shift(8446.40)

    # Recombination & forbidden emission lines
    lines = (
        # Broad Balmer
        gaussian_profile(lam_rest, l_ha, flux_ha_broad, sigma_broad)
        + gaussian_profile(lam_rest, l_hb, flux_hb_broad, sigma_broad)
        +
        # Narrow Balmer (tied to sigma_narrow)
        gaussian_profile(lam_rest, l_ha, flux_ha_narrow, sigma_narrow)
        + gaussian_profile(lam_rest, l_hb, flux_hb_narrow, sigma_narrow)
        +
        # [O III] Doublet (tied to sigma_narrow and fixed 1:2.98 flux ratio)
        gaussian_profile(lam_rest, l_oiii_4959, flux_oiii_4959, sigma_narrow)
        + gaussian_profile(lam_rest, l_oiii_5007, flux_oiii_5007, sigma_narrow)
        +
        # [O I]
        gaussian_profile(lam_rest, l_oi_8446, flux_oi_8446, sigma_narrow)
    )

    # Add the continuum model (either evaluated before as a baseline or fit simultaneously)
    mu = lines  # + continuum_model

    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)
```

#### Modified Blackbody Fitting (3.1)

fnu = Atot\*Bv(T)*(v/v0)^betaMBB
- Scale parameter Atot
- Planck function Bv(T)
- v0 = 5.45 × 1014 Hz

Strong emission lines are masked in the fit (the Balmer and Paschen
series, [O iii] doublet, O i𝜆8446, and optical to near-IR He i lines).


```python
# Step 1 - Define the blackbody & power-law models in JAX / NumPyro
import jax
import jax.numpy as jnp
import numpyro
import numpyro.distributions as dist
from numpyro.infer import MCMC, NUTS, init_to_value

H_PLANCK = 6.62607015e-34
C_LIGHT = 2.99792458e8
K_BOLTZ = 1.380649e-23
C_ANGSTROMS = 2.99792458e18  # A / s
T_REF = 26154.5
NU_0 = 5.45e14


def B_nu_jax(nu, T):
    """Planck blackbody intensity B_nu(T) in JAX.
    nu: frequency in Hz (rest-frame)
    T: temperature in Kelvin
    Returns: B_nu in W / (m^2 * Hz * sr)
    """
    # Exponent: h * nu / (k_B * T)
    x = (H_PLANCK * nu) / (K_BOLTZ * T)

    # Numerical safeguard: prevent float overflow in exp(x)
    x = jnp.clip(x, a_min=1e-6, a_max=85.0)

    # 2 * h * nu^3 / c^2
    prefactor = (2.0 * H_PLANCK * (nu**3)) / (C_LIGHT**2)

    # Use expm1(x) for numerical precision when x -> 0 (Rayleigh-Jeans limit)
    return prefactor / jnp.expm1(x)

def modified_blackbody(nu_rest, flux, flux_err):
    """Modified blackbody model from de Graaf et al. (2025).
    nu_rest: rest-frame frequency in Hz
    flux: observed flux density in W / (m^2 * Hz)
    flux_err: uncertainty in observed flux density
    Returns: None (samples from the posterior distribution of model parameters)
    """

    # a_tot scales the median flux directly (e.g. ~ 0.2 uJy)
    a_tot = numpyro.sample("a_tot", dist.LogNormal(loc=jnp.log(0.2), scale=2.0))

    # Temperature prior
    temp = numpyro.sample("temp", dist.LogUniform(low=10.0, high=1e7))
    beta_mbb = numpyro.sample("beta_mbb", dist.Uniform(low=-5.0, high=5.0))

    # Dimensionless frequency: u ~ 0.5 to 1.5
    u = nu_rest / NU_0

    # Dimensionless exponent: x = (h * nu) / (k_B * T)
    x = (T_REF / temp) * u
    x = jnp.clip(x, a_min=1e-5, a_max=85.0)

    # Dimensionless shape factor (normalized so it's ~ 1 near nu_0 at ~8000 K)
    # expm1(x) avoids subtraction cancellation when x is small
    b_shape = (u**3) / jnp.expm1(x)

    # Normalize b_shape to unity at nu_0 (u = 1) so a_tot directly tracks flux at nu_0
    x0 = jnp.clip(T_REF / temp, a_min=1e-5, a_max=85.0)
    b_norm = 1.0 / jnp.expm1(x0)
    b_profile = b_shape / b_norm

    # Modified blackbody: Profile * (nu / nu_0)^beta
    mu = a_tot * b_profile * (u**beta_mbb)

    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)


def broken_power_free_lamv(lam_rest, flux, flux_err):
    
    # Separate amplitudes on either side of the break at lambda_v
    alpha_uv = numpyro.sample("alpha_uv", dist.LogNormal(loc=jnp.log(0.1), scale=2.0))
    alpha_opt = numpyro.sample("alpha_opt", dist.LogNormal(loc=jnp.log(0.3), scale=2.0))

    # UV slope - strictly negative
    beta_uv = numpyro.sample("beta_uv", dist.TruncatedNormal(loc=-1.5, scale=1.0, high=-0.1))

    # Optical slope - strictly positive
    beta_opt = numpyro.sample("beta_opt", dist.TruncatedNormal(loc=2.0, scale=1.0, low=0.1))

    # Break wavelength (free)
    lambda_v = numpyro.sample("lambda_v", dist.Uniform(low=2500.0, high=6000.0))

    # Normalized wavelength ratio
    x = lam_rest / lambda_v

    # Discontinuous piecewise evaluation
    mu = jnp.where(
        lam_rest <= lambda_v,
        alpha_uv * (x**beta_uv),
        alpha_opt * (x**beta_opt),
    )

    # Deterministic tracking of the jump amplitude ratio (alpha_opt / alpha_uv)
    numpyro.deterministic("jump_ratio", alpha_opt / alpha_uv)
    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)

# 2. Function to execute the NUTS MCMC sampling
def generic_mcmc_fit(lam_rest, flux, flux_err, init_values, model, rng_key=42):
    """
    Executes MCMC sampling using the NUTS algorithm for a given model.
    lam_rest: rest-frame wavelength in Angstroms
    flux: observed flux density in W / (m^2 * Hz)
    flux_err: uncertainty in observed flux density
    rng_key: random seed for reproducibility
    init_values: dictionary of initial values for the model parameters
    """

    # Convert input data to JAX arrays
    lam_jax = jnp.array(lam_rest, dtype=jnp.float32)
    flux_jax = jnp.array(flux, dtype=jnp.float32)
    err_jax = jnp.array(flux_err, dtype=jnp.float32)

    # Pass init_strategy to NUTS kernel
    nuts_kernel = NUTS(
        model,
        init_strategy=init_to_value(values=init_values)
    )

    # Set up MCMC with the NUTS kernel
    mcmc = MCMC(
        nuts_kernel,
        num_warmup=250,
        num_samples=2500,
        num_chains=1
    )

    # Run MCMC sampling
    key = jax.random.PRNGKey(rng_key)
    mcmc.run(key, lam_jax, flux_jax, err_jax)
    
    mcmc.print_summary()
    posterior_samples = mcmc.get_samples()
    
    return mcmc, posterior_samples
```

```python
# Step 3 - MCMC fitting for ALL OBJECTS IN blackbody_sources
import pandas as pd
import time
from astropy.stats import sigma_clip
from astropy.io import fits

t_list = []
beta_list = []
a_list = []
t_errs = []
a_errs = []
beta_errs = []
v_list = []
v_errs = []

uv_betas = []
uv_errs = []

for source in blackbody_sources:
    # Extract source information
    dja_root = source['root']
    filename = source['file']
    z_source = source['zspec']
    print(f"\n--- Starting {filename} (z={z_source:.3f}) ---")


    # Construct the direct Amazon S3 URL
    dja_base_url = "https://s3.amazonaws.com/msaexp-nirspec/extractions/"
    full_url = f"{dja_base_url}{dja_root}/{filename}"


    # Retry loop for network resilience
    max_retries = 3
    hdul_opened = False

    for attempt in range(max_retries):
        try:
            with fits.open(full_url, cache=False) as hdul:
                spec_data = Table(hdul['SPEC1D'].data)
                sci_2d = hdul['SCI'].data
                wht_2d = hdul['WHT'].data
            hdul_opened = True
            break 
            
        except TimeoutError:
            if attempt < max_retries - 1:
                print(f"  Timeout on {filename}. Retrying in 5 seconds...")
                time.sleep(5)
            else:
                # If we run out of retries, RAISE the error so the master block catches it
                raise TimeoutError(f"S3 download failed after {max_retries} attempts.")

    # Safeguard just in case
    if not hdul_opened:
        raise RuntimeError("FITS file could not be opened for an unknown reason.")


    # Convert to rest-frame wavelength and flux density in f_lambda
    wave_obs = spec_data['wave'] * 10000 # angstroms
    wave_rest = wave_obs / (1.0 + z_source)

    detector_edge_mask = wave_obs >= 0.65
    v_lines_mask = build_line_mask(wave_rest, v_fit_emission, mask_half_width=50)
    halpha_mask = build_line_mask(wave_rest, halpha_only, mask_half_width=200)
    v_wave_limits = (wave_rest >= 1200) & (wave_rest <= 7000)

    spec_flux_flambda = spec_data['flux'] * 1e-29 * (C_ANGSTROMS / (wave_rest ** 2)) / 1e-18
    spec_noise = spec_data['err'] * 1e-29 * (C_ANGSTROMS / (wave_rest ** 2)) / 1e-18

    good_region = (wave_rest > 2000) & (wave_rest < 4000)
    median_noise = np.median(spec_noise[good_region])

    # Reject pixels whose error is an order of magnitude higher than standard
    error_mask = spec_noise < (5.0 * median_noise)

    # 2. Initialize a baseline mask for the wavelength window
    power_law_mask = detector_edge_mask & v_wave_limits & v_lines_mask & halpha_mask & error_mask

    wave_rest_masked = wave_rest[power_law_mask]
    flux_masked = spec_flux_flambda[power_law_mask]
    err_masked = spec_noise[power_law_mask]

    if np.sum(flux_masked < 0) > 0:
        print("Negative flux values present; possible contamination")


    # Estimate initial alpha as the median continuum flux near 3646 A
    near_break = (wave_rest_masked >= 3400) & (wave_rest_masked <= 3800) & (flux_masked > 0)
    alpha_start = float(np.median(flux_masked[near_break]))
    print(f"Robust starting alpha: {alpha_start:.4f}")

    # Run MCMC starting right at the measured flux level
    mcmc, samples = generic_mcmc_fit(
        wave_rest_masked, 
        flux_masked, 
        err_masked, 
        init_values={"alpha_uv": alpha_start, "alpha_opt": alpha_start, "beta_uv": -2.0, "beta_opt": 2.0, "lambda_v": 3645.0}, # start @ balmer break
        model=broken_power_free_lamv
    )

    uv_med = float(np.median(samples["beta_uv"]))
    uv_betas.append(uv_med)
    uv_errs.append(jnp.percentile(samples["beta_uv"], 84).item() - uv_med)

    # Return 84th percentile of lambda_v for clipping purposes
    lambda_v_84th = jnp.percentile(samples["lambda_v"], 84).item()
    v_med = float(np.median(samples["lambda_v"]))

    v_list.append(v_med)
    v_errs.append(lambda_v_84th - v_med)

    # Make blackbody mask
    lower_limit = max(lambda_v_84th, 4200.0)

    line_mask = build_line_mask(
        wave_rest, blackbody_emission, mask_half_width=50
    ) & build_line_mask(wave_rest, halpha_only, mask_half_width=200)

    print(np.sum(wave_rest >= lower_limit), np.sum(line_mask), np.sum(np.isfinite(spec_flux_flambda)), np.sum(np.isfinite(spec_noise)), np.sum(spec_noise > 0))
    blackbody_mask = (wave_rest >= lower_limit) & line_mask & np.isfinite(spec_flux_flambda) & np.isfinite(spec_noise) & (spec_noise > 0)

    # Extract clean numpy arrays first
    wave_clean = np.ascontiguousarray(wave_rest[blackbody_mask], dtype=np.float64)
    flux_clean = np.ascontiguousarray(
        spec_flux_flambda[blackbody_mask], dtype=np.float64
    )
    err_clean = np.ascontiguousarray(spec_noise[blackbody_mask], dtype=np.float64)

    if np.any(flux_clean < 0):
        print("Negative flux values present; possible contamination")

    # Convert rest wavelength (Angstroms) to rest frequency (Hz)
    # 1 Angstrom = 1e-10 m
    nu_clean = C_LIGHT / (wave_clean * 1e-10)

    # Calculate a reasonable scale factor a_tot using standard NumPy
    t_init = 7800.0
    beta_init = 0.0
    nu_0 = 5.45e14  # Hz

    b_nu_val = float(B_nu_jax(nu_0, t_init))

    # Median flux near 5500 Angstroms (nu ~ 5.45e14 Hz)
    near_pivot = (nu_clean > 5.0e14) & (nu_clean < 6.0e14)
    if np.any(near_pivot):
        flux_ref = float(np.median(flux_clean[near_pivot]))
    else:
        flux_ref = float(np.median(flux_clean))

    a_tot_init = max(flux_ref / b_nu_val, 1e-35)

    smooth_flux = pd.Series(flux_clean).rolling(window=9, center=True).median()
    residuals = flux_clean - smooth_flux
    clip_mask = ~sigma_clip(residuals, sigma=5.0).mask

    # Cast cleanly to JAX arrays for MCMC
    nu_rest_jax = jnp.array(nu_clean[clip_mask])
    flux_masked_jax = jnp.array(flux_clean[clip_mask])
    err_masked_jax = jnp.array(err_clean[clip_mask])

    # Run MCMC
    mcmc, samples = generic_mcmc_fit(
        nu_rest_jax,
        flux_masked_jax,
        err_masked_jax,
        init_values={"a_tot": a_tot_init, "temp": t_init, "beta_mbb": beta_init},
        model=modified_blackbody,
    )

    # Export the determined values
    a_tot_med = float(np.median(samples["a_tot"]))
    temp_med = float(np.median(samples["temp"]))
    beta_med = float(np.median(samples["beta_mbb"]))

    a_list.append(a_tot_med)
    beta_list.append(beta_med)
    t_list.append(temp_med)

    a_errs.append(jnp.percentile(samples["a_tot"], 84).item() - a_tot_med)
    beta_errs.append(jnp.percentile(samples["beta_mbb"], 84).item() - beta_med)
    t_errs.append(jnp.percentile(samples["temp"], 84).item() - temp_med)


blackbody_sources['a_tot'] = a_list
blackbody_sources['a_err'] = a_errs
blackbody_sources['b_MBB'] = beta_list
blackbody_sources['b_MBB_err'] = beta_errs
blackbody_sources['temp'] = t_list
blackbody_sources['temp_err'] = t_errs
blackbody_sources['lam_v'] = v_list
blackbody_sources['lam_v_err'] = v_errs
blackbody_sources['free_beta_uv'] = uv_betas
blackbody_sources['free_beta_uv_err'] = uv_errs

blackbody_sources.write('blackbody_sources.fits', format='fits', overwrite='True')
```
