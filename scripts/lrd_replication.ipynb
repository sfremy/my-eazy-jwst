```python
from astropy.coordinates import SkyCoord
import astropy.units as u

def find_match_by_coor(ra_target, dec_target, catalog):
    """
    Finds the closest matching target in the msaexp catalog by RA and Dec,
    and returns its S3 bucket folder root and spec.fits filename.
    """
    # 1. Convert the entire catalog coordinates into SkyCoord structures
    catalog_coords = SkyCoord(ra=catalog['ra']*u.deg, dec=catalog['dec']*u.deg)
    target_coord = SkyCoord(ra=ra_target*u.deg, dec=dec_target*u.deg)
    
    # 2. Match coordinates (finds the absolute closest index)
    idx, d2d, _ = target_coord.match_to_catalog_sky(catalog_coords)
    
    # Extract the matching row
    match_row = catalog[idx]
    
    if d2d.arcsec[0] > 1.0:
        return None
        
        
    return match_row
```

```python
# Generic MCMC for all fitting steps
import jax
import jax.numpy as jnp
import numpyro
import numpyro.distributions as dist
from numpyro.infer import MCMC, NUTS, init_to_value

def generic_mcmc_fit(
    lam_rest,
    flux,
    flux_err,
    init_values,
    model,
    *args,
    rng_key=42,
    num_warmup=250,
    num_samples=2500,
    num_chains=1,
    **kwargs,
):
    """Executes MCMC sampling using the NUTS algorithm for a given model.

    lam_rest: rest-frame wavelength in Angstroms
    flux: observed flux density in W / (m^2 * Hz) or microJansky
    flux_err: uncertainty in observed flux density
    init_values: dictionary of initial values for the model parameters
    model: NumPyro model callable
    *args, **kwargs: extra arguments passed directly to the model (e.g.
    sigma_inst_rest)
    """
    # Convert core spectrum arrays to JAX arrays
    lam_jax = jnp.array(lam_rest, dtype=jnp.float32)
    flux_jax = jnp.array(flux, dtype=jnp.float32)
    err_jax = jnp.array(flux_err, dtype=jnp.float32)

    # Pass init_strategy to NUTS kernel
    nuts_kernel = NUTS(model, init_strategy=init_to_value(values=init_values))

    # Set up MCMC with the NUTS kernel
    mcmc = MCMC(
        nuts_kernel,
        num_warmup=num_warmup,
        num_samples=num_samples,
        num_chains=num_chains,
    )

    # Run MCMC sampling, forwarding extra arguments to the model
    key = jax.random.PRNGKey(rng_key)
    mcmc.run(key, lam_jax, flux_jax, err_jax, *args, **kwargs)

    mcmc.print_summary()
    posterior_samples = mcmc.get_samples()

    return mcmc, posterior_samples
```

```python
def broken_power_fixed_lamv(lam_rest, flux, flux_err):
    
    # Separate amplitudes on either side of the break at lambda_v
    alpha_uv = numpyro.sample("alpha_uv", dist.LogNormal(loc=jnp.log(0.1), scale=2.0))
    alpha_opt = numpyro.sample("alpha_opt", dist.LogNormal(loc=jnp.log(0.3), scale=2.0))

    # UV slope - strictly negative
    beta_uv = numpyro.sample("beta_uv", dist.TruncatedNormal(loc=-1.5, scale=1.0, high=-0.1))

    # Optical slope - strictly positive
    beta_opt = numpyro.sample("beta_opt", dist.TruncatedNormal(loc=2.0, scale=1.0, low=0.1))

    # Break wavelength (fixed)
    lambda_v = 3645 

    # Normalized wavelength ratio
    x = lam_rest / lambda_v

    # Discontinuous piecewise evaluation
    mu = jnp.where(
        lam_rest <= lambda_v,
        alpha_uv * (x**beta_uv),
        alpha_opt * (x**beta_opt),
    )

    # Deterministic tracking of the jump amplitude ratio (alpha_opt / alpha_uv)
    numpyro.deterministic("jump_ratio", alpha_opt / alpha_uv)
    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)
```

#### Emission Line Fitting
- Linear continuum
- Two-component Gaussian for Balmer lines
- Single-component for [OIII] 4959/5007 doublet, OI 6300 & OI 8446
- NO [NII] doublet
- Balmer + OIII tied together
- OIII doublet flux ratio 1:2.98

Priors are in Hviding et al. (2025) except for broadline FWHM which is set to uniform distribution in range (FWHMnarrow + 100, 5000km/s)
OI prior set to [0, 2000km/s]

Also fit the non-LRD Hviding et al. emitters (37 RUBIES sources), NII doublet fixed @ 1:2.95

```python
import jax.lax as lax
import jax.numpy as jnp
import numpyro
import numpyro.distributions as dist

def integrated_gaussian(lam_grid, lam_0, flux, sigma_lam):
    """Analytic top-hat pixel-integrated Gaussian profile.

    lam_grid, lam_0, sigma_lam: all in Angstroms flux: integrated line flux
    """
    dlam = jnp.gradient(lam_grid)
    lam_lo = lam_grid - 0.5 * dlam
    lam_hi = lam_grid + 0.5 * dlam

    inv_sqrt2_sig = 1.0 / (jnp.sqrt(2.0) * sigma_lam)
    z_hi = (lam_hi - lam_0) * inv_sqrt2_sig
    z_lo = (lam_lo - lam_0) * inv_sqrt2_sig

    # Pixel-integrated flux density
    return (flux / (2.0 * dlam)) * (lax.erf(z_hi) - lax.erf(z_lo))


def get_total_sigma(lam_0, sigma_v_kms, s_lsf, sigma_inst_lam):
    """Combines intrinsic Doppler broadening and scaled instrumental LSF in

    quadrature.
    """
    sigma_kin = lam_0 * (sigma_v_kms / C_KMS)  # Angstroms
    sigma_lsf_eff = s_lsf * sigma_inst_lam  # Angstroms
    return jnp.sqrt(sigma_kin**2 + sigma_lsf_eff**2)

# Integrated flux excess: (peak - continuum) * approx_fwhm (e.g. ~40-60 A in PRISM)
def estimate_line_flux(lam_grid, flux_grid, lam_line, cont_level, dlam=50.0):
    region = (lam_grid >= lam_line - dlam) & (lam_grid <= lam_line + dlam)
    if np.any(region):
        excess = np.maximum(flux_grid[region] - cont_level, 0.0)
        # Numerical trapezoid sum across the line feature
        return float(np.trapezoid(excess, lam_grid[region]))
    return 0.5

def get_prism_sigma_inst_rest(lam_rest_angstroms, z):
    """Calculates rest-frame instrumental Gaussian dispersion (in Angstroms)

    for a point source observed with JWST/NIRSpec PRISM/CLEAR.
    """
    lam_rest_angstroms = np.asarray(lam_rest_angstroms, dtype=float)
    lam_obs_um = (lam_rest_angstroms * (1.0 + z)) / 1e4

    # Point-source resolution curve R(lambda)
    r_prism = 35.0 + 42.5 * lam_obs_um + 2.8 * (lam_obs_um**2)

    # Convert resolving power to rest-frame standard deviation in Angstroms
    # FWHM_rest = lam_rest / R -> sigma_rest = lam_rest / (2.35482 * R)
    sigma_rest = lam_rest_angstroms / (2.35482 * r_prism)
    return sigma_rest

def joint_emission_lines(lam_rest, flux, flux_err, sigma_inst_rest):
    # -------------------------------------------------------------
    # 1. Kinematics and Instrumental Priors
    # -------------------------------------------------------------
    sigma_narrow = numpyro.sample(
        "w_narrow", dist.Uniform(low=0.0, high=750.0)
    )  # km/s
    max_delta = 2500.0 - (sigma_narrow + 100.0)

    # Dynamically construct sigma_broad
    delta_w = numpyro.sample(
        "delta_w_broad", dist.Uniform(low=0.0, high=max_delta)
    )
    sigma_broad = numpyro.deterministic("w_broad", sigma_narrow + 100.0 + delta_w)

    # Freely varying OI width
    sigma_oi = numpyro.sample(
            "sigma_oi", dist.Uniform(low=0.0, high=2000.0)
    )  # km/s

    s_lsf = numpyro.sample(
        "s_lsf", dist.TruncatedNormal(loc=1.2, scale=0.1, low=0.9, high=1.5)
    )
    v_off = numpyro.sample("v_off", dist.Normal(loc=0.0, scale=300.0))  # km/s

    # Doppler shifted line centers
    lam_shift = lambda lam0: lam0 * (1.0 + v_off / C_KMS)

    l_ha = lam_shift(6564.61)
    l_hb = lam_shift(4862.68)
    l_oiii_4959 = lam_shift(4960.30)
    l_oiii_5007 = lam_shift(5008.24)
    l_oi_8446 = lam_shift(8446.40)

    # -------------------------------------------------------------
    # 2. Quadrature Widths in Angstroms
    # -------------------------------------------------------------
    # sigma_inst_rest is interpolated to the respective rest wavelengths:
    # e.g., inst_ha = sigma_inst_rest at 6565 A (~20 - 30 A)
    sig_ha_b = get_total_sigma(
        l_ha, sigma_broad, s_lsf, sigma_inst_rest["ha"]
    )
    sig_hb_b = get_total_sigma(
        l_hb, sigma_broad, s_lsf, sigma_inst_rest["hb"]
    )

    sig_ha_n = get_total_sigma(
        l_ha, sigma_narrow, s_lsf, sigma_inst_rest["ha"]
    )
    sig_hb_n = get_total_sigma(
        l_hb, sigma_narrow, s_lsf, sigma_inst_rest["hb"]
    )
    sig_oiii_4959 = get_total_sigma(
        l_oiii_4959, sigma_narrow, s_lsf, sigma_inst_rest["oiii_4959"]
    )
    sig_oiii_5007 = get_total_sigma(
        l_oiii_5007, sigma_narrow, s_lsf, sigma_inst_rest["oiii_5007"]
    )
    sig_oi_8446 = get_total_sigma(
        l_oi_8446, sigma_oi, s_lsf, sigma_inst_rest["oi_8446"]
    )

    # -------------------------------------------------------------
    # 3. Flux Amplitudes
    # -------------------------------------------------------------
    flux_ha_broad = numpyro.sample(
        "flux_ha_broad", dist.HalfNormal(scale=10.0)
    )
    flux_ha_narrow = numpyro.sample(
        "flux_ha_narrow", dist.HalfNormal(scale=5.0)
    )
    flux_hb_broad = numpyro.sample(
        "flux_hb_broad", dist.HalfNormal(scale=5.0)
    )
    flux_hb_narrow = numpyro.sample(
        "flux_hb_narrow", dist.HalfNormal(scale=2.0)
    )

    flux_oiii_5007 = numpyro.sample(
        "flux_oiii_5007", dist.HalfNormal(scale=5.0)
    )
    flux_oiii_4959 = flux_oiii_5007 / 2.98
    flux_oi_8446 = numpyro.sample("flux_oi_8446", dist.HalfNormal(scale=2.0))

    # -------------------------------------------------------------
    # 4. Synthesize Profiles
    # -------------------------------------------------------------
    lines = (
        integrated_gaussian(lam_rest, l_ha, flux_ha_broad, sig_ha_b)
        + integrated_gaussian(lam_rest, l_hb, flux_hb_broad, sig_hb_b)
        + integrated_gaussian(lam_rest, l_ha, flux_ha_narrow, sig_ha_n)
        + integrated_gaussian(lam_rest, l_hb, flux_hb_narrow, sig_hb_n)
        + integrated_gaussian(
            lam_rest, l_oiii_4959, flux_oiii_4959, sig_oiii_4959
        )
        + integrated_gaussian(
            lam_rest, l_oiii_5007, flux_oiii_5007, sig_oiii_5007
        )
        + integrated_gaussian(lam_rest, l_oi_8446, flux_oi_8446, sig_oi_8446)
    )

    # -------------------------------------------------------------
    # 5. Handle Continuum
    # -------------------------------------------------------------
    # 1. Blue Window (H-beta + [O III]): ~4619 - 5259 A
    theta_blue = numpyro.sample(
        "theta_blue", dist.Uniform(low=-jnp.pi / 2, high=jnp.pi / 2)
    )
    height_blue = numpyro.sample("height_blue", dist.Uniform(low=1e-4, high=3.0))
    cont_blue = height_blue + jnp.tan(theta_blue) * ((lam_rest - 4940.0) / 320.0)

    # 2. Red Window (H-alpha): ~6236 - 6893 A
    theta_red = numpyro.sample(
        "theta_red", dist.Uniform(low=-jnp.pi / 2, high=jnp.pi / 2)
    )
    height_red = numpyro.sample("height_red", dist.Uniform(low=1e-4, high=3.0))
    cont_red = height_red + jnp.tan(theta_red) * ((lam_rest - 6565.0) / 330.0)

    # 3. NIR Window ([O I] 8446): ~8024 - 8869 A
    theta_oi = numpyro.sample(
        "theta_oi", dist.Uniform(low=-jnp.pi / 2, high=jnp.pi / 2)
    )
    height_oi = numpyro.sample("height_oi", dist.Uniform(low=1e-4, high=3.0))
    cont_oi = height_oi + jnp.tan(theta_oi) * ((lam_rest - 8446.0) / 420.0)

    # -------------------------------------------------------------
    # Piecewise Continuum Assignment
    # -------------------------------------------------------------
    piecewise_continuum = jnp.where(
        lam_rest < 5800.0,
        cont_blue,
        jnp.where(lam_rest < 7500.0, cont_red, cont_oi),
    )

    mu = lines + piecewise_continuum
    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)
```

```python
from astropy.io import fits
import time

# Extract source information
dja_root = source['root']
filename = source['file']
z_source = source['zspec']
print(f"\n--- Starting {filename} (z={z_source:.3f}) ---")

line_wavelengths = {
    "ha": 6564.61,
    "hb": 4862.68,
    "oiii_4959": 4960.30,
    "oiii_5007": 5008.24,
    "oi_8446": 8446.40,
}

# Precompute rest-frame instrumental widths (in Angstroms) for your target's redshift
sigma_inst_rest = {
    key: float(get_prism_sigma_inst_rest(wave, z_source))
    for key, wave in line_wavelengths.items()
}

# Example diagnostic printout for a z ~ 5.5 source:
# ha: ~18-22 A, hb: ~15-18 A, oi: ~22-26 A
for line_name, sig_val in sigma_inst_rest.items():
    print(
        f"{line_name:>10}: rest sigma_inst = {sig_val:6.2f} A  (FWHM ~ {sig_val * 2.355:5.1f} A)"
    )

# Construct the direct Amazon S3 URL
dja_base_url = "https://s3.amazonaws.com/msaexp-nirspec/extractions/"
full_url = f"{dja_base_url}{dja_root}/{filename}"


# Retry loop for network resilience
max_retries = 3
hdul_opened = False

for attempt in range(max_retries):
    try:
        with fits.open(full_url, cache=False) as hdul:
            spec_data = Table(hdul['SPEC1D'].data)
            sci_2d = hdul['SCI'].data
            wht_2d = hdul['WHT'].data
        hdul_opened = True
        break 
        
    except TimeoutError:
        if attempt < max_retries - 1:
            print(f"  Timeout on {filename}. Retrying in 5 seconds...")
            time.sleep(5)
        else:
            # If we run out of retries, RAISE the error so the master block catches it
            raise TimeoutError(f"S3 download failed after {max_retries} attempts.")

# Safeguard just in case
if not hdul_opened:
    raise RuntimeError("FITS file could not be opened for an unknown reason.")


# Convert to rest-frame wavelength and flux density in f_lambda
wave_obs = spec_data['wave'] * 10000 # angstroms
wave_rest = wave_obs / (1.0 + z_source)
spec_flux_flambda = spec_data['flux'] * 1e-29 * (C_ANGSTROMS / (wave_rest ** 2)) / 1e-18
spec_noise = spec_data['err'] * 1e-29 * (C_ANGSTROMS / (wave_rest ** 2)) / 1e-18

# Define window boundaries (rest frame)
blue_window = (wave_rest >= 4619.0) & (wave_rest <= 5259.0)
red_window  = (wave_rest >= 6236.0) & (wave_rest <= 6893.0)
oi_window   = (wave_rest >= 8024.0) & (wave_rest <= 8869.0)

# Master window mask (union of all regions of interest)
fit_window_mask = blue_window | red_window | oi_window

# Optional: apply detector edge and basic quality masks
fit_window_mask = fit_window_mask & (wave_obs >= 6500.0) & np.isfinite(spec_flux_flambda) & np.isfinite(spec_noise)

# Slice arrays to pass to NumPyro
lam_fit   = wave_rest[fit_window_mask]
flux_fit  = spec_flux_flambda[fit_window_mask]
err_fit   = spec_noise[fit_window_mask]

# --- A. Continuum Heights and Slopes ---
# Measure median continuum level in the line-free shoulders (+/- 3500 km/s masked)
# Blue window: around 4700-4800 A and 5100-5200 A
c_blue_init = float(np.median(flux_fit[(lam_fit >= 4650) & (lam_fit <= 4750)]))
# Red window: around 6300-6400 A and 6800-6880 A
c_red_init  = float(np.median(flux_fit[(lam_fit >= 6300) & (lam_fit <= 6400)]))
# OI window: around 8100-8200 A and 8700-8800 A
c_oi_init   = float(np.median(flux_fit[(lam_fit >= 8100) & (lam_fit <= 8200)]))

# Safeguard against negative dips/noise
c_blue_init = max(c_blue_init, 1e-3)
c_red_init  = max(c_red_init,  1e-3)
c_oi_init   = max(c_oi_init,   1e-3)

# --- B. Empirical Line Flux Estimates ---

ha_tot   = max(estimate_line_flux(lam_fit, flux_fit, 6564.61, c_red_init, dlam=70.0), 0.1)
hb_tot   = max(estimate_line_flux(lam_fit, flux_fit, 4862.68, c_blue_init, dlam=50.0), 0.05)
oiii_tot = max(estimate_line_flux(lam_fit, flux_fit, 5008.24, c_blue_init, dlam=50.0), 0.05)
oi_tot   = max(estimate_line_flux(lam_fit, flux_fit, 8446.40, c_oi_init,   dlam=50.0), 0.02)

init_values = {
    # -------------------------------------------------------------
    # Kinematics & Instrumental
    # -------------------------------------------------------------
    # Narrow-line FWHM: center of prior range (0, 750) km/s -> ~350 km/s
    "w_narrow": 350.0,
    
    # Excess broad width: place w_broad around ~1800 - 2000 km/s
    # If using delta_w_broad: (e.g., 350 + 100 + 1400 = 1850 km/s)
    "delta_w_broad": 1400.0,
    # (If using uniform fraction u_broad: set to 0.5)
    # "u_broad": 0.5,
    
    # LSF scale factor: set to the prior mode/mean (1.2)
    "s_lsf": 1.2,
    
    # Systemic velocity / redshift offset: start at zero
    "v_off": 0.0,

    # -------------------------------------------------------------
    # Local Linear Continua
    # -------------------------------------------------------------
    # Heights anchored at the measured shoulder levels
    "height_blue": c_blue_init,
    "height_red":  c_red_init,
    "height_oi":   c_oi_init,
    
    # Angles: start flat (0.0 rad -> slope = 0)
    "theta_blue": 0.0,
    "theta_red":  0.0,
    "theta_oi":   0.0,

    # -------------------------------------------------------------
    # Emission Line Fluxes
    # -------------------------------------------------------------
    # In LRDs, the broad component typically accounts for ~60-80% of Balmer flux
    "flux_ha_broad":  ha_tot * 0.70,
    "flux_ha_narrow": ha_tot * 0.30,
    
    "flux_hb_broad":  hb_tot * 0.70,
    "flux_hb_narrow": hb_tot * 0.30,
    
    # [O III] 5007 and [O I] 8446 single line fluxes
    "flux_oiii_5007": oiii_tot,
    "flux_oi_8446":   oi_tot,
}

# Run MCMC
mcmc, samples = generic_mcmc_fit(
    lam_fit,
    flux_fit,
    err_fit,
    init_values,
    joint_emission_lines,
    sigma_inst_rest,  # <-- *args forwards this to the 4th parameter of joint_emission_lines
    rng_key=42,
)

# 2. Plot 3-window fit and residuals
plot_mcmc_fit(lam_fit, flux_fit, err_fit, samples, sigma_inst_rest)

# 3. Plot MCMC parameter covariance corner plot
plot_corner_diagnostics(samples)
```

```python
import corner

# test plot
def evaluate_components(
    lam, params, sigma_inst_rest, return_decomposed=False
):
    """Evaluates individual lines and piecewise continua given a parameter dictionary

    lam: 1D rest-frame wavelength array (in Angstroms) params: dictionary of
    scalar parameter values
    """
    v_off = params["v_off"]
    lam_shift = lambda lam0: lam0 * (1.0 + v_off / C_KMS)

    # Reconstruct FWHM -> dispersion
    # If using delta_w_broad:
    w_narrow = params["w_narrow"]
    w_broad = (
        params.get("w_broad")
        if "w_broad" in params
        else (w_narrow + 100.0 + params["delta_w_broad"])
    )

    sig_v_n = w_narrow / 2.35482
    sig_v_b = w_broad / 2.35482
    s_lsf = params["s_lsf"]

    # Quadrature dispersion calculation
    def calc_sigma(l0, sig_v, sig_inst):
        sig_kin = l0 * (sig_v / C_KMS)
        return np.sqrt(sig_kin**2 + (s_lsf * sig_inst) ** 2)

    # Pixel integration helper (numpy version)
    from scipy.special import erf

    def np_integrated_gaussian(l_grid, l0, flux, sig):
        dlam = np.gradient(l_grid)
        z_hi = (l_grid + 0.5 * dlam - l0) / (np.sqrt(2.0) * sig)
        z_lo = (l_grid - 0.5 * dlam - l0) / (np.sqrt(2.0) * sig)
        return (flux / (2.0 * dlam)) * (erf(z_hi) - erf(z_lo))

    # Line evaluations
    l_ha = lam_shift(6564.61)
    l_hb = lam_shift(4862.68)
    l_o4959 = lam_shift(4960.30)
    l_o5007 = lam_shift(5008.24)
    l_oi = lam_shift(8446.40)

    ha_b = np_integrated_gaussian(
        lam,
        l_ha,
        params["flux_ha_broad"],
        calc_sigma(l_ha, sig_v_b, sigma_inst_rest["ha"]),
    )
    ha_n = np_integrated_gaussian(
        lam,
        l_ha,
        params["flux_ha_narrow"],
        calc_sigma(l_ha, sig_v_n, sigma_inst_rest["ha"]),
    )
    hb_b = np_integrated_gaussian(
        lam,
        l_hb,
        params["flux_hb_broad"],
        calc_sigma(l_hb, sig_v_b, sigma_inst_rest["hb"]),
    )
    hb_n = np_integrated_gaussian(
        lam,
        l_hb,
        params["flux_hb_narrow"],
        calc_sigma(l_hb, sig_v_n, sigma_inst_rest["hb"]),
    )
    o5007 = np_integrated_gaussian(
        lam,
        l_o5007,
        params["flux_oiii_5007"],
        calc_sigma(l_o5007, sig_v_n, sigma_inst_rest["oiii_5007"]),
    )
    o4959 = np_integrated_gaussian(
        lam,
        l_o4959,
        params["flux_oiii_5007"] / 2.98,
        calc_sigma(l_o4959, sig_v_n, sigma_inst_rest["oiii_4959"]),
    )
    oi8446 = np_integrated_gaussian(
        lam,
        l_oi,
        params["flux_oi_8446"],
        calc_sigma(l_oi, sig_v_n, sigma_inst_rest["oi_8446"]),
    )

    # Continua evaluations
    c_blue = params["height_blue"] + np.tan(params["theta_blue"]) * (
        (lam - 4940.0) / 320.0
    )
    c_red = params["height_red"] + np.tan(params["theta_red"]) * (
        (lam - 6565.0) / 330.0
    )
    c_oi = params["height_oi"] + np.tan(params["theta_oi"]) * (
        (lam - 8446.0) / 420.0
    )

    continuum = np.where(lam < 5800.0, c_blue, np.where(lam < 7500.0, c_red, c_oi))
    total_model = (
        ha_b + ha_n + hb_b + hb_n + o5007 + o4959 + oi8446 + continuum
    )

    if return_decomposed:
        return {
            "total": total_model,
            "continuum": continuum,
            "broad": ha_b + hb_b,
            "narrow": ha_n + hb_n + o5007 + o4959 + oi8446,
        }
    return total_model

def plot_mcmc_fit(
    lam_fit,
    flux_fit,
    err_fit,
    samples,
    sigma_inst_rest,
    num_draws=100,
    save_path=None,
):
    # Median parameters
    median_params = {k: float(np.median(v)) for k, v in samples.items()}
    best_decomp = evaluate_components(
        lam_fit, median_params, sigma_inst_rest, return_decomposed=True
    )

    # Posterior uncertainty band
    rng = np.random.default_rng(42)
    sample_indices = rng.choice(len(samples["s_lsf"]), size=num_draws, replace=False)
    draws = np.array([
        evaluate_components(
            lam_fit,
            {k: float(v[i]) for k, v in samples.items()},
            sigma_inst_rest,
        )
        for i in sample_indices
    ])
    p16, p84 = np.percentile(draws, [16, 84], axis=0)

    # Window definitions: (name, [min_lambda, max_lambda])
    windows = [
        (r"H$\beta$ + [O III]", [4619.0, 5259.0]),
        (r"H$\alpha$", [6236.0, 6893.0]),
        (r"[O I] $\lambda8446$", [8024.0, 8869.0]),
    ]

    fig = plt.figure(figsize=(16, 6))
    gs = fig.add_gridspec(2, 3, height_ratios=[3, 1], hspace=0.08, wspace=0.22)

    for col_idx, (win_name, (w_min, w_max)) in enumerate(windows):
        ax_top = fig.add_subplot(gs[0, col_idx])
        ax_res = fig.add_subplot(gs[1, col_idx], sharex=ax_top)

        m = (lam_fit >= w_min) & (lam_fit <= w_max)
        if not np.any(m):
            continue

        l_sub = lam_fit[m]
        f_sub = flux_fit[m]
        e_sub = err_fit[m]

        # Top panel: Data and decomposition
        ax_top.step(
            l_sub, f_sub, where="mid", color="black", lw=1.2, label="PRISM Data"
        )
        ax_top.fill_between(
            l_sub,
            f_sub - e_sub,
            f_sub + e_sub,
            step="mid",
            color="gray",
            alpha=0.25,
            label=r"Rescaled $\sigma$",
        )

        ax_top.plot(
            l_sub,
            best_decomp["total"][m],
            color="#D95F02",
            lw=1.8,
            label="Total Model",
        )
        ax_top.fill_between(
            l_sub, p16[m], p84[m], color="#D95F02", alpha=0.2, label="16–84th %"
        )

        ax_top.plot(
            l_sub,
            best_decomp["continuum"][m],
            color="gray",
            ls="--",
            lw=1.2,
            label="Continuum",
        )
        ax_top.plot(
            l_sub,
            best_decomp["continuum"][m] + best_decomp["broad"][m],
            color="#7570B3",
            ls=":",
            lw=1.4,
            label="Broad Lines",
        )
        ax_top.plot(
            l_sub,
            best_decomp["continuum"][m] + best_decomp["narrow"][m],
            color="#1B9E77",
            ls="-.",
            lw=1.4,
            label="Narrow Lines",
        )

        ax_top.set_title(win_name, fontsize=12, fontweight="semibold")
        ax_top.tick_params(labelbottom=False)
        ax_top.set_ylabel(r"$f_\nu\ [\mu\mathrm{Jy}]$")
        if col_idx == 0:
            ax_top.legend(fontsize=8, loc="upper left", frameon=False)

        # Bottom panel: Normalized residuals (Pull)
        residuals = (f_sub - best_decomp["total"][m]) / e_sub
        ax_res.step(l_sub, residuals, where="mid", color="black", lw=0.9)
        ax_res.axhline(0, color="gray", ls="--", lw=1)
        ax_res.axhline(2, color="red", ls=":", lw=0.8, alpha=0.7)
        ax_res.axhline(-2, color="red", ls=":", lw=0.8, alpha=0.7)
        ax_res.set_ylim(-3.5, 3.5)
        ax_res.set_ylabel(r"$(f - \mu)/\sigma$")
        ax_res.set_xlabel(r"$\lambda_{\mathrm{rest}}\ [\mathrm{\AA}]$")

    if save_path:
        plt.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.show()

def plot_corner_diagnostics(samples, save_path=None):
    # Compute deterministic w_broad if stored as delta_w_broad
    w_narrow = np.array(samples["w_narrow"])
    if "w_broad" in samples:
        w_broad = np.array(samples["w_broad"])
    else:
        w_broad = w_narrow + 100.0 + np.array(samples["delta_w_broad"])

    corner_data = np.vstack([
        w_narrow,
        w_broad,
        np.array(samples["s_lsf"]),
        np.array(samples["flux_ha_broad"]),
        np.array(samples["flux_ha_narrow"]),
        np.array(samples["flux_oiii_5007"]),
    ]).T

    labels = [
        r"$w_{\mathrm{narrow}}\ [\mathrm{km/s}]$",
        r"$w_{\mathrm{broad}}\ [\mathrm{km/s}]$",
        r"$s_{\mathrm{LSF}}$",
        r"$F(\mathrm{H}\alpha_{\mathrm{broad}})$",
        r"$F(\mathrm{H}\alpha_{\mathrm{narrow}})$",
        r"$F(\mathrm{[O\ III]}\ 5007)$",
    ]

    fig = corner.corner(
        corner_data,
        labels=labels,
        quantiles=[0.16, 0.50, 0.84],
        show_titles=True,
        title_fmt=".2f",
        title_kwargs={"fontsize": 10},
        color="#2B5C8F",
        hist_kwargs={"density": True},
        smooth=1.0,
    )

    if save_path:
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.show()
```

#### Balmer Break Integration

To measure the Balmer break strength we integrate the spectrum in two tophat filters with wavelength ranges [3620,3720] Å and [4000,4100] Å using pyphot (Fouesneau 2025), and compute the flux density ratio (in fν)1. Error bars are estimated using 500 random Gaussian draws from the error spectrum.

#### Modified Blackbody Fitting (3.1)

fnu = Atot\*Bv(T)*(v/v0)^betaMBB
- Scale parameter Atot
- Planck function Bv(T)
- v0 = 5.45 × 1014 Hz

Strong emission lines are masked in the fit (the Balmer and Paschen
series, [O iii] doublet, O i𝜆8446, and optical to near-IR He i lines).


```python
# Step 1 - Define the blackbody & power-law models in JAX / NumPyro
def B_nu_jax(nu, T):
    """Planck blackbody intensity B_nu(T) in JAX.
    nu: frequency in Hz (rest-frame)
    T: temperature in Kelvin
    Returns: B_nu in W / (m^2 * Hz * sr)
    """
    # Exponent: h * nu / (k_B * T)
    x = (H_PLANCK * nu) / (K_BOLTZ * T)

    # Numerical safeguard: prevent float overflow in exp(x)
    x = jnp.clip(x, a_min=1e-6, a_max=85.0)

    # 2 * h * nu^3 / c^2
    prefactor = (2.0 * H_PLANCK * (nu**3)) / (C_LIGHT**2)

    # Use expm1(x) for numerical precision when x -> 0 (Rayleigh-Jeans limit)
    return prefactor / jnp.expm1(x)

def modified_blackbody(nu_rest, flux, flux_err):
    """Modified blackbody model from de Graaf et al. (2025).
    nu_rest: rest-frame frequency in Hz
    flux: observed flux density in W / (m^2 * Hz)
    flux_err: uncertainty in observed flux density
    Returns: None (samples from the posterior distribution of model parameters)
    """

    # a_tot scales the median flux directly (e.g. ~ 0.2 uJy)
    a_tot = numpyro.sample("a_tot", dist.LogNormal(loc=jnp.log(0.2), scale=2.0))

    # Temperature prior
    temp = numpyro.sample("temp", dist.LogUniform(low=10.0, high=1e7))
    beta_mbb = numpyro.sample("beta_mbb", dist.Uniform(low=-5.0, high=5.0))

    # Dimensionless frequency: u ~ 0.5 to 1.5
    u = nu_rest / NU_0

    # Dimensionless exponent: x = (h * nu) / (k_B * T)
    x = (T_REF / temp) * u
    x = jnp.clip(x, a_min=1e-5, a_max=85.0)

    # Dimensionless shape factor (normalized so it's ~ 1 near nu_0 at ~8000 K)
    # expm1(x) avoids subtraction cancellation when x is small
    b_shape = (u**3) / jnp.expm1(x)

    # Normalize b_shape to unity at nu_0 (u = 1) so a_tot directly tracks flux at nu_0
    x0 = jnp.clip(T_REF / temp, a_min=1e-5, a_max=85.0)
    b_norm = 1.0 / jnp.expm1(x0)
    b_profile = b_shape / b_norm

    # Modified blackbody: Profile * (nu / nu_0)^beta
    mu = a_tot * b_profile * (u**beta_mbb)

    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)


def broken_power_free_lamv(lam_rest, flux, flux_err):
    
    # Separate amplitudes on either side of the break at lambda_v
    alpha_uv = numpyro.sample("alpha_uv", dist.LogNormal(loc=jnp.log(0.1), scale=2.0))
    alpha_opt = numpyro.sample("alpha_opt", dist.LogNormal(loc=jnp.log(0.3), scale=2.0))

    # UV slope - strictly negative
    beta_uv = numpyro.sample("beta_uv", dist.TruncatedNormal(loc=-1.5, scale=1.0, high=-0.1))

    # Optical slope - strictly positive
    beta_opt = numpyro.sample("beta_opt", dist.TruncatedNormal(loc=2.0, scale=1.0, low=0.1))

    # Break wavelength (free)
    lambda_v = numpyro.sample("lambda_v", dist.Uniform(low=2500.0, high=6000.0))

    # Normalized wavelength ratio
    x = lam_rest / lambda_v

    # Discontinuous piecewise evaluation
    mu = jnp.where(
        lam_rest <= lambda_v,
        alpha_uv * (x**beta_uv),
        alpha_opt * (x**beta_opt),
    )

    # Deterministic tracking of the jump amplitude ratio (alpha_opt / alpha_uv)
    numpyro.deterministic("jump_ratio", alpha_opt / alpha_uv)
    numpyro.sample("obs", dist.Normal(loc=mu, scale=flux_err), obs=flux)

# 2. Function to execute the NUTS MCMC sampling
def generic_mcmc_fit(lam_rest, flux, flux_err, init_values, model, rng_key=42):
    """
    Executes MCMC sampling using the NUTS algorithm for a given model.
    lam_rest: rest-frame wavelength in Angstroms
    flux: observed flux density in W / (m^2 * Hz)
    flux_err: uncertainty in observed flux density
    rng_key: random seed for reproducibility
    init_values: dictionary of initial values for the model parameters
    """

    # Convert input data to JAX arrays
    lam_jax = jnp.array(lam_rest, dtype=jnp.float32)
    flux_jax = jnp.array(flux, dtype=jnp.float32)
    err_jax = jnp.array(flux_err, dtype=jnp.float32)

    # Pass init_strategy to NUTS kernel
    nuts_kernel = NUTS(
        model,
        init_strategy=init_to_value(values=init_values)
    )

    # Set up MCMC with the NUTS kernel
    mcmc = MCMC(
        nuts_kernel,
        num_warmup=250,
        num_samples=2500,
        num_chains=1
    )

    # Run MCMC sampling
    key = jax.random.PRNGKey(rng_key)
    mcmc.run(key, lam_jax, flux_jax, err_jax)
    
    mcmc.print_summary()
    posterior_samples = mcmc.get_samples()
    
    return mcmc, posterior_samples
```
